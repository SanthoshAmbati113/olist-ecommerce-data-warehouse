```sql
WITH category_sales AS (
    SELECT
        p.product_category,
        SUM(f.price) AS category_sales

    FROM `e-commerce`.ecommerce_gold.fact_order_items f

    JOIN `e-commerce`.ecommerce_gold.dim_products p
        ON f.product_sk = p.product_sk

    GROUP BY p.product_category
)

SELECT
    product_category,

    ROUND(category_sales, 2) AS category_sales,

    ROUND(
        100.0 * category_sales /
        NULLIF(SUM(category_sales) OVER (), 0),
        2
    ) AS sales_contribution_pct

FROM category_sales

ORDER BY category_sales DESC;
```
