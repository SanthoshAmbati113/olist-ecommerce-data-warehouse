```sql
WITH monthly_sales AS (
    SELECT
        d.year_month,
        SUM(f.price) AS monthly_sales

    FROM `e-commerce`.ecommerce_gold.fact_order_items f

    JOIN `e-commerce`.ecommerce_gold.dim_date d
        ON f.purchase_date_key = d.date_key

    GROUP BY d.year_month
),

ranked AS (
    SELECT
        *,
        RANK() OVER (
            ORDER BY monthly_sales DESC
        ) AS highest_rank,

        RANK() OVER (
            ORDER BY monthly_sales ASC
        ) AS lowest_rank

    FROM monthly_sales
)

SELECT
    year_month,
    ROUND(monthly_sales, 2) AS monthly_sales,

    CASE
        WHEN highest_rank = 1 THEN 'Highest Sales'
        WHEN lowest_rank = 1 THEN 'Lowest Sales'
    END AS sales_status

FROM ranked

WHERE highest_rank = 1
   OR lowest_rank = 1

ORDER BY monthly_sales DESC;
```
