```sql
WITH monthly_metrics AS (
    SELECT
        d.year,
        d.month,
        d.year_month,

        SUM(f.price) AS sales,

        COUNT(DISTINCT f.order_id) AS orders

    FROM `e-commerce`.ecommerce_gold.fact_order_items f

    JOIN `e-commerce`.ecommerce_gold.dim_date d
        ON f.purchase_date_key = d.date_key

    GROUP BY d.year, d.month, d.year_month
)

SELECT
    year_month,

    ROUND(sales, 2) AS sales,

    orders,

    ROUND(
        sales / NULLIF(orders, 0),
        2
    ) AS average_order_value

FROM monthly_metrics

ORDER BY year, month;
```
