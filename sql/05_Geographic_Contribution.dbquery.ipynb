```sql
WITH state_sales AS (
    SELECT
        g.state,

        SUM(f.price) AS state_sales,

        COUNT(DISTINCT f.order_id) AS total_orders

    FROM `e-commerce`.ecommerce_gold.fact_order_items f

    JOIN `e-commerce`.ecommerce_gold.dim_geography g
        ON f.geography_sk = g.geography_sk

    GROUP BY g.state
)

SELECT
    state,

    ROUND(state_sales, 2) AS state_sales,

    total_orders,

    ROUND(
        100.0 * state_sales /
        NULLIF(SUM(state_sales) OVER (), 0),
        2
    ) AS sales_contribution_pct

FROM state_sales

ORDER BY state_sales DESC;
```
